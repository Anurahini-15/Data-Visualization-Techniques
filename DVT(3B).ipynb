```python
import pandas as pd

file_path = r"C:\Users\palan\Downloads\archive (1)\credit_card_fraud_10k.csv"

df = pd.read_csv(file_path)

df.head()
```

```python
df.info()
```

```python
print(df.isnull().sum())
```

```python
df = df.dropna()

print("Missing values removed successfully")
```

```python
print("Duplicate Records:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("Duplicate records removed successfully")
```

```python
print(df.describe())
```

```python
print(df.columns)
```

```python

df['amount_normalized'] = (df['amount'] - df['amount'].min()) / (df['amount'].max() - df['amount'].min())

print(df[['amount', 'amount_normalized']].head())
```

```python
df['merchant_category_encoded'] = df['merchant_category'].astype('category').cat.codes

print(df[['merchant_category', 'merchant_category_encoded']].head())
```

```python
df.to_csv(r"C:\Users\palan\Downloads\cleaned_credit_card_fraud.csv", index=False)

print("Cleaned dataset exported successfully")
```
