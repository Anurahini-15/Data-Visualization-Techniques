```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_excel(r"C:\Users\palan\Downloads\credit_card_fraud.xlsx")

print(df.head())
print(df.columns)
```

```python
data = df["is_fraud"].value_counts()

plt.bar(data.index, data.values)

plt.title("Fraud vs Non-Fraud Transactions")
plt.xlabel("Fraud Status")
plt.ylabel("Number of Transactions")

plt.show()
```

```python
fraud_count = df["is_fraud"].value_counts()

plt.pie(
    fraud_count,
    labels=["Not Fraud", "Fraud"],
    autopct='%1.1f%%'
)

plt.title("Fraud Transaction Distribution")
plt.show()
```

```python
plt.boxplot(df["amount"])

plt.title("Transaction Amount")
plt.ylabel("Amount")

plt.show()
```

```python
data = df.groupby("transaction_hour")["is_fraud"].sum()

plt.plot(data.index, data.values)

plt.title("Fraud Transactions by Hour")
plt.xlabel("Transaction Hour")
plt.ylabel("Number of Fraud Transactions")

plt.show()
```
