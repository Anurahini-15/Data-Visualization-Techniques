```python
import pandas as pd

df = pd.read_csv(r"C:\Users\palan\OneDrive\Documents\employee.csv")
print(df)
```

```python
print(df.isnull())
```

```python
print(df.isnull().sum())
```

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())

df["Salary"] = df["Salary"].fillna(df["Salary"].mean())

print(df)
```

```python
print(df.duplicated())

df.drop_duplicates(inplace=True)

print(df)
```

```python
Q1 = df["Salary"].quantile(0.25)

Q3 = df["Salary"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR

upper = Q3 + 1.5 * IQR

outliers = df[(df["Salary"] < lower) | (df["Salary"] > upper)]

print(outliers)
```

```python
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\palan\OneDrive\Documents\employee.csv")
print(df)
```

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())

print(df)
```
